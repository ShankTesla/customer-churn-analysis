```python
import numpy as np
import pandas as pd
import random
```

```python
class EDA:

    def missing_values(self, df):
        # 0. Utility variables and functions
        cols = df.columns 
        missing_dict = {}
        percentage = lambda x: float(round((x/len(df)) * 100,2))

        # 1. Save missing values to dict
        for column in cols:
            missing_dict[column] = {"num_of_missing_values": int(df[column].isnull().sum()), "percentage": percentage(df[column].isnull().sum())}
        
        # 2. Return
        return missing_dict
    

    def duplicate_rows(self, df):
        # 0. Utility variables and functions
        duplicate_rows = {}
        percentage = lambda x: float(round((x/len(df)) * 100,2))
        duplicates = int(df.duplicated().sum())

        # 1. Save duplicates to dictionary
        duplicate_rows['duplicate_count'] = duplicates
        duplicate_rows['duplicate_percent'] = percentage(duplicates)

        # 2. Return
        return duplicate_rows
    
    def data_types(self, df):
        # 1. Return
        return df.dtypes
    
    def min_value(self, df):
        # 1. Return
        return df.min()
    
    def max_value(self, df):
        # 1. Return
        return df.max()
    
    def unique_counts(self, df):
        # 0. Utility variables and functions
        cols = df.columns
        uniq_counts = {}

        for column in cols:
            uniq_counts[column] = df[column].value_counts()
        return uniq_counts


    def data_audit(self, df):
        # 0. Utility variables if any

        # 1. missing values and its percentages per column
        missing_values = self.missing_values(df)
        
        # 2. Duplicate row detection
        duplicate_rows = self.duplicate_rows(df)

        # 3. Data type list 
        data_types = self.data_types(df)

        # 4. Value distribution checks (min/max/unique_counts)
        mins = self.min_value(df)
        max = self.max_value(df)
        unique_counts = self.unique_counts(df)

        # 5. results
        print("="*10+"Data Audit"+"="*10)
        print("==Missing values and it's percentages Audit==")
        display(missing_values)
        print("==Duplicatw rows audit==")
        display(duplicate_rows)
        print("==Data type Audit==")
        display(data_types)
        print("==Min, Max, Unique counts audit==")
        print("Min")
        display(mins)
        print("Max")
        display(max)
        print("Unique Counts")
        display(unique_counts)
        print("="*10+"End of Audit"+"="*10)
```

```python
# 1. load data
df = pd.read_csv("../data/raw/customer_churn_raw.csv")

# 2. create a class object
eda = EDA()

# 3. pass the data and call a method
eda.data_audit(df)
```
